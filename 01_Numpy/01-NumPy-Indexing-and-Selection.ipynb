## 대괄호 인덱싱과 선택
배열의 하나 또는 일부 요소를 선택하는 가장 간단한 방법은 파이썬 리스트와 매우 유사합니다:

```python
#Get a value at an index
arr[8]
```

```python
#Get values in a range
arr[1:5]
```

```python
#Get values in a range
arr[0:5]
```

```python
arr[:5]
```

```python
arr[5:]
```

## 브로드캐스팅

NumPy 배열은 브로드캐스팅 기능 때문에 일반 파이썬 리스트와 다릅니다. 리스트에서는 같은 크기와 형태의 새로운 부분으로만 일부를 재할당할 수 있습니다. 예를 들어 리스트의 처음 5개 요소를 새 값으로 바꾸려면 새로운 5개짜리 리스트를 넘겨야 합니다. NumPy 배열에서는 단일 값을 더 큰 값 집합에 브로드캐스트(확장)할 수 있습니다:

```python
#Setting a value with index range (Broadcasting)
arr[0:5]=100

#Show
arr
```

```python
# Reset array, we'll see why I had to reset in  a moment
arr = np.arange(0,11)

#Show
arr
```

```python
#Important notes on Slices
slice_of_arr = arr[0:6]

#Show slice
slice_of_arr
```

```python
#Change Slice
slice_of_arr[:]=99

#Show Slice again
slice_of_arr
```

이제 변경사항이 원본 배열에도 반영되는 것을 확인하세요!

```python
arr
```

데이터가 복사되는 것이 아니라 원본 배열의 뷰(view)입니다! 이로 인해 메모리 문제를 피할 수 있습니다!

```python
#To get a copy, need to be explicit
arr_copy = arr.copy()

arr_copy
```

## 2차원 배열(행렬) 인덱싱

일반 형식은 **arr_2d[row][col]** 또는 **arr_2d[row,col]** 입니다. 명확성을 위해 쉼표 표기법을 사용하는 것을 권장합니다.

```python
arr_2d = np.array(([5,10,15],[20,25,30],[35,40,45]))

#Show
arr_2d
```

```python
#Indexing row
arr_2d[1]
```

```python
# Format is arr_2d[row][col] or arr_2d[row,col]

# Getting individual element value
arr_2d[1][0]
```

```python
# Getting individual element value
arr_2d[1,0]
```

```python
# 2D array slicing

#Shape (2,2) from top right corner
arr_2d[:2,1:]
```

```python
#Shape bottom row
arr_2d[2]
```

```python
#Shape bottom row
arr_2d[2,:]
```

```python
arr_2d.shape
```
