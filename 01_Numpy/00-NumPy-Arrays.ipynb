# NumPy 배열

NumPy 배열은 이 강의 전반에서 NumPy를 사용하는 주요 방식입니다. NumPy 배열에는 크게 두 가지 형태가 있습니다: 벡터와 행렬입니다. 벡터는 엄격히 1차원(1D) 배열이고, 행렬은 2차원(2D)입니다(단, 행렬은 한 행 또는 한 열만 가질 수 있습니다).

이제 NumPy 배열을 만드는 방법을 살펴보면서 소개를 시작하겠습니다.

## NumPy 배열 생성

### 파이썬 리스트에서

리스트 또는 리스트의 리스트를 바로 변환하여 배열을 만들 수 있습니다:

```python
my_list = [1,2,3]
my_list
```

```python
np.array(my_list)
```

```python
my_matrix = [[1,2,3],[4,5,6],[7,8,9]]
my_matrix
```

```python
np.array(my_matrix)
```

## 내장 메서드

배열을 생성하는 여러 가지 내장 기능이 있습니다.

### arange

주어진 구간 내에서 균등한 간격의 값을 반환합니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.arange.html)]

```python
np.arange(0,10)
```

```python
np.arange(0,11,2)
```

### zeros와 ones

0 또는 1로 채워진 배열을 생성합니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.zeros.html)]

```python
np.zeros(3)
```

```python
np.zeros((5,5))
```

```python
np.ones(3)
```

```python
np.ones((3,3))
```

```python
np.ones((5,5)) + 5
```

```python
np.ones(4) / 100
```

### linspace 
지정된 구간에 걸쳐 균등한 간격으로 숫자를 반환합니다. [[참고](https://www.numpy.org/devdocs/reference/generated/numpy.linspace.html)]

```python
np.linspace(0,10,3)
```

```python
np.linspace(0,5,20)
```

<font color=green>`.linspace()`는 종료값을 포함합니다. 일반적인 분수를 얻으려면 항목 수를 늘리세요:</font>

```python
np.linspace(0,5,21)
```

```python
np.linspace(0, 10, 6)
```

### eye

단위 행렬(identity matrix)을 생성합니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.eye.html)]

```python
np.eye(4)
```

```python
np.eye(10)
```

## 난수
NumPy에는 난수 배열을 생성하는 여러 방법이 있습니다:

### rand
지정한 형태의 배열을 만들고 균등 분포 [0, 1)에서 샘플로 채웁니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.rand.html)]

```python
np.random.rand(2)
```

```python
np.random.rand(5,5)
```

```python
np.random.randint(1, 45, size=6)
```

### randn

표준 정규 분포(σ = 1)에서 샘플(또는 샘플들)을 반환합니다. **rand**가 균등 분포인 것과 달리, 0에 가까운 값이 더 자주 나타납니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.randn.html)]

```python
np.random.randn(2)
```

```python
np.random.randn(5, 5)
```

### randint

`low`(포함)에서 `high`(미포함) 사이의 정수를 무작위로 반환합니다.  [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.randint.html)]

```python
np.random.randint(1,100)
```

```python
np.random.randint(1,100,10)
```

### seed

난수 상태를 설정하여 동일한 '무작위' 결과를 재현할 수 있습니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.random.seed.html)]

```python
np.random.seed(42)
np.random.rand(4)
```

```python
np.random.seed(42)
np.random.rand(4)
```

## 배열 속성 및 메서드

배열에서 자주 쓰이는 유용한 속성과 메서드를 살펴보겠습니다:

```python
arr = np.arange(25)
ranarr = np.random.randint(0,50,10)
```

```python
arr
```

```python
ranarr
```

```python
ranarr.max()
```

```python
ranarr.min()
```

```python
ranarr.argmax() # index of max value
```

```python
ranarr.argmin() # index of min value
```

## Reshape

같은 데이터를 가지면서 새로운 모양의 배열을 반환합니다. [[참고](https://docs.scipy.org/doc/numpy-1.15.0/reference/generated/numpy.reshape.html)]

```python
arr.reshape(5,5)
```

### max, min, argmax, argmin

이 메서드들은 최댓값이나 최솟값을 찾을 때 유용합니다. 또한 argmin이나 argmax를 사용하면 해당 값들의 인덱스 위치를 알 수 있습니다.

```python
ranarr
```

```python
ranarr.max()
```

```python
ranarr.argmax()
```

```python
ranarr.min()
```

```python
ranarr.argmin()
```
