#### 11. 2010년 인구조사와 2017년 인구 추정치 사이의 증감률을 계산하는 열을 추가하세요.

```python
# 여기에 코드 작성
df['PercentChange'] = 100 * (df['2017PopEstimate'] - df['2010Census']) / df['2010Census']
# 결과를 확인하려면 이것을 사용하세요
df.head()
```

```python
# 여기에 작성하지 마세요
```

<strong>보너스: 2010년 인구조사와 2017년 인구 추정치 사이의 예상 증감률이 가장 높은 주(State)는 어디인가요?</strong><br>이 문제는 PercentChange를 다시 계산해야 하므로 여러 줄의 코드가 필요합니다.

```python
# 주(State)별로 그룹화하여 인구 합산
states = df.groupby('State').sum()

# 2010~2017년 인구 변화율(%) 계산
states['PercentChange'] = 100 * (states['2017PopEstimate'] - states['2010Census']) / states['2010Census']

# 변화율 기준 내림차순 정렬 후 상위 5개 주 출력
states.sort_values(by='PercentChange', ascending=False).head()
```
