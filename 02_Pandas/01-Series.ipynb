# Series 소개
pandas에서 가장 먼저 배울 주요 데이터 타입은 Series입니다. Pandas를 불러오고 Series 객체를 살펴보겠습니다.

Series는 NumPy 배열과 매우 비슷합니다. 실제로 NumPy 배열 객체를 기반으로 만들어져 있습니다. 하지만 NumPy 배열과 달리 Series는 축 레이블을 가질 수 있어서 단순한 숫자 위치뿐 아니라 레이블로도 인덱싱할 수 있습니다. 또한 숫자 데이터만 담을 필요가 없으며, 임의의 Python 객체도 저장할 수 있습니다.

몇 가지 예제로 이 개념을 살펴보겠습니다:

```python
import numpy as np
import pandas as pd
```

## Series 만들기

리스트, NumPy 배열, 딕셔너리를 Series로 변환할 수 있습니다:

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d = {'a':10,'b':20,'c':30}
```

### 리스트 사용

```python
pd.Series(data=my_list)
```

```python
pd.Series(data=my_list,index=labels)
```

```python
pd.Series(my_list,labels)
```

### NumPy 배열 사용

```python
pd.Series(arr)
```

```python
pd.Series(arr,labels)
```

### 딕셔너리 사용

```python
pd.Series(d)
```

### Series 안의 데이터

pandas Series는 다양한 객체 타입을 담을 수 있습니다:

```python
pd.Series(data=labels)
```

```python
# 함수도 넣을 수 있습니다 (실제로 이렇게 사용할 일은 거의 없습니다)
pd.Series([sum,print,len])
```

## 인덱스 사용하기

Series를 사용하는 핵심은 인덱스를 이해하는 것입니다. Pandas는 이 인덱스 이름이나 숫자를 활용해 정보를 빠르게 조회할 수 있게 해줍니다. 해시 테이블이나 딕셔너리처럼 동작합니다.

이제 Series에서 정보를 가져오는 몇 가지 예를 살펴보겠습니다. ser1과 ser2라는 두 개의 Series를 만들어 보겠습니다:

```python
ser1 = pd.Series([1,2,3,4],index = ['USA', 'Germany','USSR', 'Japan'])                                   
```

```python
ser1
```

```python
ser2 = pd.Series([1,2,5,4],index = ['USA', 'Germany','Italy', 'Japan'])                                   
```

```python
ser2
```

```python
ser1['USA']
```

연산도 인덱스를 기준으로 수행됩니다:

```python
ser1 + ser2  # 한쪽에만 있는 경우는 NaN으로 처리됩니다.
```

여기서는 잠시 멈추고, Series 개념을 확장하는 DataFrames로 넘어가겠습니다!
