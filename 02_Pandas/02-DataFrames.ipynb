# DataFrames 소개

DataFrame은 pandas에서 가장 핵심적으로 사용되는 구조이며 R 프로그래밍 언어에서 직접적인 영감을 받았습니다. DataFrame은 여러 Series 객체가 같은 인덱스를 공유하도록 묶여 있는 형태라고 생각할 수 있습니다. pandas를 사용해 이 주제를 살펴보겠습니다!

```python
import pandas as pd
import numpy as np
```

```python
from numpy.random import randn
np.random.seed(101)
```

```python
df = pd.DataFrame(randn(5,4),index='A B C D E'.split(),columns='W X Y Z'.split())
```

```python
df
```

## 선택과 인덱싱

DataFrame에서 데이터를 가져오는 여러 방법을 살펴보겠습니다.

```python
df['W']
```

```python
# 열 이름 목록 전달
df[['W','Z']]
```

```python
# SQL 문법 (권장하지 않음!)
df.W
```

DataFrame의 열은 결국 Series입니다

```python
type(df['W'])
```

### 새 열 만들기:

```python
df['new'] = df['W'] + df['Y']
```

```python
df
```

### 열 제거하기

```python
df.drop('new',axis=1)
```

```python
# inplace=True를 지정하지 않으면 원본은 바뀌지 않습니다!
df
```

```python
df.drop('new',axis=1,inplace=True)
```

```python
df
```

행도 같은 방식으로 삭제할 수 있습니다:

```python
df.drop('E',axis=0)
```

### 행 선택하기

```python
df.loc['A']
```

또는 라벨 대신 위치를 기준으로 선택할 수 있습니다.

```python
df.iloc[2]
```

### 행과 열의 일부 선택하기

```python
df.loc['B','Y']
```

```python
df.loc[['A','B'],['W','Y']]
```

### 조건 선택

pandas의 중요한 기능 중 하나는 대괄호 표기법을 이용한 조건 선택이며, 이는 NumPy와 매우 비슷합니다:

```python
df
```

```python
df>0
```

```python
df[df>0]
```

```python
df[df['W']>0]
```

```python
df[df['W']>0]['Y']
```

```python
df[df['W']>0][['Y','X']]
```

두 개의 조건에는 괄호와 함께 | 와 & 를 사용할 수 있습니다:

```python
df[(df['W']>0) & (df['Y'] > 1)]
```

## 인덱스에 대한 추가 설명

인덱스를 재설정하거나 다른 값으로 설정하는 방법 등, 인덱싱의 몇 가지 추가 기능을 살펴보겠습니다. 또한 계층적 인덱스도 함께 다룹니다!

```python
df
```

```python
# 기본 0,1...n 인덱스로 재설정
df.reset_index()
```

```python
newind = 'CA NY WY OR CO'.split()
```

```python
df['States'] = newind
```

```python
df
```

```python
df.set_index('States')
```

```python
df
```

```python
df.set_index('States',inplace=True)
```

```python
df
```
