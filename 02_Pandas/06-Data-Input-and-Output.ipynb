# 데이터 입력과 출력

이 노트북은 입력과 출력에 대한 참고 코드입니다. pandas는 다양한 파일 형식을 pd.read_ 계열 메서드로 읽을 수 있습니다. 가장 자주 쓰이는 데이터 형식을 살펴보겠습니다:

```python
import numpy as np
import pandas as pd
```

## CSV
쉼표로 구분된 값(Comma Separated Values) 파일은 필드를 쉼표로 구분하는 텍스트 파일입니다.<br>
강의에 포함된 가상 환경을 사용하지 않는다면 <tt>xlrd</tt>와 <tt>openpyxl</tt>를 설치해야 할 수 있습니다.<br>

### CSV 입력

```python
df = pd.read_csv('example.csv')
df
```

### CSV 출력

```python
df.to_csv('example.csv',index=False)
```

## Excel
Pandas는 MS Excel 파일을 읽고 쓸 수 있습니다. 하지만 이것은 데이터만 가져오며, 수식이나 이미지는 가져오지 않습니다. 이미지나 매크로가 포함된 파일은 <tt>.read_excel()</tt> 메서드가 중단될 수 있습니다.

### Excel 입력

```python
pd.read_excel('Excel_Sample.xlsx',sheet_name='Sheet1')
```

### Excel 출력

```python
df.to_excel('Excel_Sample.xlsx',sheet_name='Sheet1')
```
